```python
import requests

# Define a function to geocode a single place
def geocode_place(place):
    base_url = "https://nominatim.openstreetmap.org/search"
    params = {
        "q": place,
        "format": "json",
    }

    response = requests.get(base_url, params=params)
    data = response.json()

    if data:
        # Extract latitude and longitude from the first result
        lat = data[0]["lat"]
        lon = data[0]["lon"]
        return lat, lon
    else:
        return None

# List of places you want to geocode
places = ["Bengaluru"]


lat=[]
lon=[]
# Geocode each place in the list
for place in places:
    coordinates = geocode_place(place)
    if coordinates:
        print(f"{place}: Latitude {coordinates[0]}, Longitude {coordinates[1]}")
        lat.append(coordinates[0]),lon.append(coordinates[1])
    else:
        print(f"Could not geocode {place}")
```

```python
data={"Lat":lat,"Lon":lon}
import pandas as pd
df=pd.DataFrame(data)
df.to_csv("UK1st.csv")
```
