```python
import folium

# Create a world map object
world_map = folium.Map(location=[0, 0], zoom_start=2)

# Add a marker for each country you want to highlight
folium.Marker([56.1304, -106.3468], tooltip='Canada').add_to(world_map)
folium.Marker([37.0902, -95.7129], tooltip='USA').add_to(world_map)
folium.Marker([23.6345, -102.5528], tooltip='Mexico').add_to(world_map)

# Save the map to an HTML file
world_map.save('countries_map.html')
```
