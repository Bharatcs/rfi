```python
import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
# Load a world map shapefile from geopandas.datasets
world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))
crs_sys='EPSG:4326'
world = world.to_crs(crs_sys)

# Create a Pandas DataFrame with country codes and values
data=pd.read_csv('/home/pratush/Downloads/Fin_HDI.csv')

# Merge the world map with the DataFrame
world = world.merge(data, left_on='name', right_on='Country')

# Create a Choropleth map
world.plot(column='Value', cmap='YlGn', legend=True)

# Show the map
plt.show()

```
