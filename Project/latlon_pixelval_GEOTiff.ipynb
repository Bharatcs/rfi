```python

#imports#
import rasterio as ras
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
```

```python
# Open GeoTIFF file#
img = ras.open('/media/sf_Shared_Ubuntu/MAPS/GPW_2020_1deg.tif')
data = img.read(1)

# Function to get pixel value at given longitude and latitude#
def getval(lon, lat):
    idx = img.index(lon, lat, precision=1E-6)
    return data[idx]
print(np.shape(data))
hor=data[:1,:]
ver=data[:,:1]
print(np.where(np.min(data)))
```

```python

table_data = []

# Iterate through latitude and longitude values#
for lat in range(-89,90):
    for lon in range(-179,180):
        pixel_value = getval(lon, lat)
        table_data.append({'Longitude': lon, 'Latitude': lat, 'Pixel Value': pixel_value})

# fill a dataframe with table_data#
table_df = pd.DataFrame(table_data)
table_df["Latitude"]=table_df["Latitude"]


  
#print(table_df)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', 1)
 
# All dataframes hereafter reflect these changes.
display(table_df)
```
