```python
import pytz
from datetime import datetime
from dateutil.relativedelta import relativedelta
from skyfield.api import load, wgs84, EarthSatellite
import numpy as np
import pandas as pd
```

```python
file=open('../landsat_archive')
content=file.read()
ts = load.timescale()
eph=load('de421.bsp')
```

## slice one year of TLEs

```python
lineone=[]
linetwo=[]
num=2000
year=np.zeros(num)
month=np.zeros(num)
days=np.zeros((num))
hour=np.zeros((num))
minute=np.zeros((num))
second=np.zeros((num))
for i in range(2000):
    line11=(content[((2*i)*70):((2*i)*70)+69])
    line22=(content[((2*i+1)*70):((2*i+1)*70)+69])
    sate=EarthSatellite(line11,line22,'Landsat7',ts)
#     print(sate)
    lineone.append(line11)
    linetwo.append(line22)
    year=sate.epoch.utc.year
    month=sate.epoch.utc.month
    day=sate.epoch.utc.day
    days[i]+=day
    if year==2000.0:
        if month==4.0:
            if day==15.0:
                   break
    
    
    
```

## Get index of first TLE of every day of the year

```python
uniq_idx=[]
uniq_idx.append(0)
for i in range(len(lineone)):
    if i>0:
        if days[i]!=days[i-1]:
            uniq_idx.append(i)
        
   
```

## Ground track for one year with changing TLE

```python
subpoints=[]
sunlit=[]
for i in uniq_idx:
    line1=lineone[i]
    line2=linetwo[i]
    satellite = EarthSatellite(line1, line2, 'Landsat-8', ts)
    print(satellite)
    tz = pytz.timezone('UTC')
    st=satellite.epoch.utc
    dt=tz.fromutc(datetime(st.year,st.month,st.day,st.hour,st.minute,0))
    t0 = ts.utc(dt)
    t1 = ts.utc(dt + relativedelta(hours=24))
    timescales = ts.linspace(t0,t1, 10000)
    # calculate the subpoints.

    geocentrics = satellite.at(timescales)
    sunlit.append(geocentrics.is_sunlit(eph))
    subpoints.append(wgs84.subpoint_of(geocentrics))
    # altitude=satellite.epoch.utc
#     print(altitude)
    
    
```
