```python
import numpy as np
import healpy as hp
import pandas as pd
import rasterio 
```

```python

# Open the GeoTIFF file
file_path = '/home/pratush/5kRasters/USA_shpraster.tif'
src = rasterio.open(file_path)

# Read the raster data as a numpy array
data = src.read(1)  # Assuming a single-band raster

# Find pixels with non-zero values
non_zero_pixels = np.where(data != 0)

# Get the coordinates (latitude, longitude) of non-zero pixels
lat=[]
lon=[]
for y, x in zip(*non_zero_pixels):
    lonval, latval = src.xy(y, x)  # Get the longitude and latitude of the pixel
    lat.append(latval)
    lon.append(lonval)
```

```python
fulldf=pd.read_csv('/home/pratush/Yogen_STARFIRE/STARFIRE_2_backup/Project/database/USA.csv')
```

```python
fulon=fulldf['Longitude in degrees'].to_numpy()
fulat=fulldf['Latitude in degrees'].to_numpy()
```

```python
goodpix=hp.ang2pix(128,lon,lat,lonlat=True)
```

```python
goodlon=[]
goodlat=[]
for i in range(len(fulon)):
    pix=hp.ang2pix(128,fulon[i],fulat[i],lonlat=True)
    if pix in goodpix:
        goodlon.append(fulon[i])
        goodlat.append(fulat[i])
```

```python
df=pd.DataFrame({'Lon':goodlon,'Lat':goodlat})
```
