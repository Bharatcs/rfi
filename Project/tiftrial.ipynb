```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import rasterio
from rasterio.plot import show
from rasterio import features





```

```python
# Load GDP data from a CSV file
gdp_data = pd.read_csv('/home/pratush/Downloads/Fin_HDI.csv')

# Load country boundaries shapefile
world = gpd.read_file('/home/pratush/Downloads/ne_110m_admin_0_countries/ne_110m_admin_0_countries.shp')

# Merge the GDP data with country boundaries
world = world.merge(gdp_data, left_on='SOVEREIGNT', right_on='Country', how='left')
```

```python
# Create a choropleth map
fig, ax = plt.subplots(figsize=(10, 10))
world.plot(column='SOVEREIGNT', cmap='inferno',  ax=ax)

# Customize map appearance, add title, labels, etc.
```
