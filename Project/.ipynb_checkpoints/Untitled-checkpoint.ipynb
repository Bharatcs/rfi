```python
import geopandas as gpd

# Open the GeoJSON file
gdf = gpd.read_file("/home/pratush/Yogen_STARFIRE/output.geojson")

# Extract data or perform spatial operations
print(gdf.head())  # Display the first few rows of the GeoDataFrame

# Access specific attributes
print(gdf['geometry'])
```

```python

# Initialize an empty list to store the coordinates
coordinates = []

# Iterate through the geometries, handling multi-part geometries
for geom in gdf['geometry']:
    if geom.geom_type == 'Polygon':
        # For individual polygons, get interior rings
        for interior in geom.interiors:
            coordinates.append(list(interior.coords))
    elif geom.geom_type == 'MultiPolygon':
        # For multi-part polygons, iterate over each part and get interior rings
        for part in geom.geoms:
            for interior in part.interiors:
                coordinates.append(list(interior.coords))

# Display the first few interior coordinates
# Replace 5 with the number of features you want to display
```

```python
lon_lat_tuples=coordinates
# Initialize empty arrays for lon and lat
lon_array = []
lat_array = []

# Iterate through the list of coordinates
for coordinate_set in coordinates:
    lon = [lon for lon, _ in coordinate_set]  # Extract lon (the first value in each tuple)
    lat = [lat for _, lat in coordinate_set]  # Extract lat (the second value in each tuple)
    
    lon_array.append(lon)
    lat_array.append(lat)

# Display the resulting arrays
print("Longitude Arrays:", lon_array)
print("Latitude Arrays:", lat_array)
```
