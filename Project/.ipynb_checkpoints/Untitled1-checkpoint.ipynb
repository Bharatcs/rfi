```python
from PIL import Image, ImageFilter

# Open an already existing image
imageObject = Image.open("/home/pratush/hires_fmscan.tif")

# Apply sharp filter
sharpened1 = imageObject.filter(ImageFilter.SHARPEN)
sharpened2 = sharpened1.filter(ImageFilter.SHARPEN)

# Save the sharpened images in TIFF format
sharpened1.save("/home/pratush/sharpened1.tif")
sharpened2.save("/home/pratush/sharpened2.tif")
```
