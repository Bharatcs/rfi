```python
import numpy as np
import matplotlib.pyplot as plt
import tifffile as tiff
import healpy as hp

# Read the tif file
image = tiff.imread('/media/sf_Shared_Ubuntu/MAPS/GPW_2020_1deg.tif')
image_array = np.array(image)

# HEALPix scheme for image array
theta = np.linspace(0, np.pi, num=image_array.shape[0])
phi = np.linspace(-np.pi, np.pi, num=image_array.shape[1])

# Calculate the desired nside based on the number of pixels in the image
desired_npix = len(image_array)
nside = 1
while 12 * nside ** 2 < desired_npix:
    nside *= 2

# Round down to the nearest valid nside
nside //= 2

# Allocate HEALPix Map to image array
healpix_map = np.zeros(hp.nside2npix(nside), dtype=np.double)

# Loop through each pixel and assign the value from the image_array to the corresponding HEALPix pixel
for i in range(len(theta)):
    for j in range(len(phi)):
        pixel_index = hp.ang2pix(nside, theta[i], phi[j])
        healpix_map[pixel_index] = image_array

# Use ud_grade to upsample the HEALPix map to the desired nside
desired_healpix_map = hp.pixelfunc.ud_grade(healpix_map, nside_out=16)

# Plot the HEALPix map
hp.mollview(desired_healpix_map, cmap="inferno", xsize=3000, flip="geo", norm="hist", min=0)
plt.show()

# Check the maximum and minimum values in the HEALPix map
print(np.max(desired_healpix_map))
print(np.min(desired_healpix_map))
```
