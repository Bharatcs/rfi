## Imports

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("/home/pratush/Downloads/Fin_HDI.csv")
world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))
```

```python
world_data = world.merge(data, left_on='name', right_on='Country')


world_data.plot(column='Value', legend=True)
plt.title('World Map with Data')

plt.show()
```

```python
country_data = world_data[world_data['name'] == 'India']
print(country_data[['name', 'Value', 'geometry']])
```

```python
centroid = country_data['geometry'].centroid
longitude, latitude = centroid.x.values[0], centroid.y.values[0]
print(f'Longitude: {longitude}, Latitude: {latitude}')
```
