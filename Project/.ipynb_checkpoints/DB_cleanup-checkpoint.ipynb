```python
import numpy as np
import pandas as pd
import glob
```

```python
csv_files = glob.glob('/home/pratush/Yogen_STARFIRE/To_fixDB/*.csv')





# --Create an empty dataframe to store the combined data-- #
combined_df = pd.DataFrame()

# Loop through each CSV file and append its contents to the combined dataframe
for csv_file in (csv_files):
    df_temp = pd.read_csv(csv_file)
    df_temp=df_temp.dropna()
    lat=df_temp['Latitude in degrees']
    arcmin=(lat*100)%100
    arcsec=(arcmin*100)%100
    arcmin=arcmin.astype(int)
    lat=lat.astype(int)
    lat+=(arcmin/60)+(arcsec/3600)
    df_temp['Latitude in degrees']=lat
    lon=df_temp['Longitude in degrees']
    arcmin1=(lon*100)%100
    arcsec1=(arcmin1*100)%100
    arcmin1=arcmin1.astype(int)
    lon=lon.astype(int)
    lon+=(arcmin1/60)+(arcsec1/3600)
    df_temp['Longitude in degrees']=lon
    df_temp.to_csv(csv_file)

#     df_temp['Latitude in degrees']=lat
#     df_temp['Longitude in degrees']=lon
```
