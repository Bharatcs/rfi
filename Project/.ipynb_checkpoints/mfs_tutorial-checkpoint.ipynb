## Compute scalar MFs, CMT and Alpha

$$
W_{0}(\nu)=\int_{\mathcal{S}^{2}}{\rm d}a\,\Theta(u-\nu) \\
W_{1}(\nu)=\int_{\mathcal{S}^{2}}{\rm d}a\,\delta (u-\nu)\,|\nabla u| \\
W_{2}(\nu)=\frac{1}{2\pi}\int_{\mathcal{S}^{2}}{\rm d}a\,\delta (u-\nu)\,|\nabla u|\, \kappa \\
W_{1}^{0,2}=(\nu)=\int_{\mathcal{S}^{2}}{\rm d}a\,\delta (u-\nu)\,\frac{1}{|\nabla u|}\mathcal{M}
$$

where $\mathcal{M}$ is given as,
$$
\mathcal{M}=
\begin{pmatrix}
u_{;2}^{2} & u_{;1}u_{;2} \\
u_{;1}u_{;2} & u_{;1}^{2} 
\end{pmatrix}
$$

For numerical purpose, $\delta(u-\nu)$ is approximated as,
$$
\delta(u-\nu)\approx\frac{1}{\Delta\nu}
$$
if $u\in[\nu-\Delta\nu/2,\nu+\Delta\nu/2]$ and zero, otherwise.


```python
def compute_mfs(umap,mask,nu0,nu1,dnu,snu):

    npix=len(umap)
    nside=hp.npix2nside(npix)
    lmax=3*nside
    nu=np.arange(nu0, nu1, dnu)
    a,b=min(umap),max(umap) 
    
    map_der=hp.alm2map_der1(hp.map2alm(umap,lmax),nside,lmax)
    map_der1=hp.alm2map_der1(hp.map2alm(map_der[1],lmax),nside,lmax)
    map_der2=hp.alm2map_der1(hp.map2alm(map_der[2],lmax),nside,lmax)

    theta=hp.pix2ang(nside,np.arange(npix))[0]
    grad=((map_der[1]**2)+(map_der[2]**2))
    
    ut=map_der[1]
    up=map_der[2]
    sn=np.sin(theta) 
    cot=1/(np.tan(theta))

    utt=map_der1[1] 
    utp=(map_der1[2])-cot*up 
    upp=map_der2[2]+cot*ut

    B1=np.sqrt(grad)
    B2=((2*ut*up*utp)-((ut**2)*(upp))-((up**2)*(utt)))/(grad) 
    
    tot_nu=len(nu) 
    W0=np.zeros_like(nu) 
    W1=np.zeros_like(nu)
    W2=np.zeros_like(nu) 

    W1_02=np.zeros([tot_nu,3])
    alpha=np.zeros_like(nu) 
    
    for inu in range(len(nu)):
        
        pix0=np.where(np.logical_and(umap>nu[inu],mask>snu)) 
        pix1=np.where((umap>=(nu[inu]-0.5*dnu))&(umap<=(nu[inu]+0.5*dnu))&(mask>snu))
        count1=np.sum((umap>=(nu[inu]-0.5*dnu))&(umap<=(nu[inu]+0.5*dnu))&(mask>snu)) 
        masksum=np.sum(mask[np.where(mask>snu)])
    
        W0[inu]=np.sum(mask[pix0])/masksum
        W1[inu]=0.25*(1/dnu)*np.sum(mask[pix1]*B1[pix1])/masksum
        W2[inu]=(1/6.28)*(((1/dnu)*np.sum(mask[pix1]*B2[pix1]))/masksum)  
    
        W1_02[inu,0]=0.25*(1/dnu)*np.sum(mask[pix1]*((ut[pix1]*ut[pix1])/B1[pix1]))/masksum 
        W1_02[inu,1]=-0.25*(1/dnu)*np.sum(mask[pix1]*((ut[pix1]*up[pix1])/B1[pix1]))/masksum
        W1_02[inu,2]=0.25*(1/dnu)*np.sum(mask[pix1]*((up[pix1]*up[pix1])/B1[pix1]))/masksum
    
        #alpha
        det=W1_02[inu,0]*W1_02[inu,2]-(W1_02[inu,1]**2)
        trace=W1_02[inu,0]+W1_02[inu,2] 
        E1=(trace/2)-np.sqrt((trace**2)/4-det)
        E2=(trace/2)+np.sqrt((trace**2)/4-det)
    
        if ((nu[inu]>=a)&(nu[inu]<=b)):  
            alpha[inu]=E1/E2  

    
    return nu,W0,W1,W2,W1_02,alpha
```

```python
nu0=-4.0
nu1=4.2
dnu=0.2
snu=0.9

nu,W0,W1,W2,W1_02,alpha=compute_mfs(map2,mask,nu0,nu1,dnu,snu)
```

```python
fig, axs = plt.subplots(2,3,figsize=(15,10)) 

fig.subplots_adjust(left=0.08,right=0.98,wspace=0.30,top=0.79,bottom=0.14)  

axs[0,0].plot(nu,W0,c='red',lw=2,label='CMB') 
axs[0,0].grid()
axs[0,0].set_ylabel(r'$W_{0}$',fontsize=20) 
axs[0,0].set_xlabel(r'$\nu$',fontsize=20) 

axs[0,1].plot(nu,W1,c='red',lw=2,label='CMB') 
axs[0,1].grid()
axs[0,1].set_ylabel(r'$W_{1}$',fontsize=20) 
axs[0,1].set_xlabel(r'$\nu$',fontsize=20) 

axs[0,2].plot(nu,W2,c='red',lw=2,label='CMB') 
axs[0,2].grid()
axs[0,2].set_ylabel(r'$W_{2}$',fontsize=20) 
axs[0,2].set_xlabel(r'$\nu$',fontsize=20) 

axs[1,0].plot(nu,W1_02[:,0],c='red',lw=2,label='CMB') 
axs[1,0].grid()
axs[1,0].set_ylabel(r'$W_{1}^{0,2}(1,1)$',fontsize=20) 
axs[1,0].set_xlabel(r'$\nu$',fontsize=20) 

axs[1,1].plot(nu,W1_02[:,2],c='red',lw=2,label='CMB') 
axs[1,1].grid()
axs[1,1].set_ylabel(r'$W_{1}^{0,2}(2,2)$',fontsize=20) 
axs[1,1].set_xlabel(r'$\nu$',fontsize=20) 

axs[1,2].plot(nu,alpha,c='red',lw=2,label='CMB') 
axs[1,2].grid()
axs[1,2].set_ylabel(r'$\alpha$',fontsize=20) 
axs[1,2].set_xlabel(r'$\nu$',fontsize=20) 

```

### Gaussian expectations for MFs

$$
\langle W_{0}(\nu)\rangle=\frac{1}{2}\,{\rm erfc\Big(\frac{\nu}{\sqrt{2}}\Big)}\\
\langle W_{1}(\nu)\rangle\propto \frac{1}{r_{c}}\,{\rm exp}\Big(-\frac{\nu^{2}}{2}\Big) \\
\langle W_{2}(\nu)\rangle\propto \frac{1}{r_{c}^{2}}\nu\,{\rm exp}\Big(-\frac{\nu^{2}}{2}\Big)
$$
where $r_{c}=\sigma_{0}/\sigma_{1}$

```python
W0_analyt=np.zeros_like(W0)
W1_analyt=np.zeros_like(W1)
W2_analyt=np.zeros_like(W2)

s0,s1=sigma0,sigma1
for inu in range(len(nu)):
    W0_analyt[inu]=0.5*(erfc(nu[inu]/np.sqrt(2))) 
    W1_analyt[inu]=(1/(8*np.sqrt(2)))*(s1/s0)*(np.exp(-(nu[inu]**2)/2))
    W2_analyt[inu]=(1/(4*np.sqrt(2)*(np.pi**(1.5))))*(s1/s0)*(s1/s0)*(nu[inu])*(np.exp(-(nu[inu]**2)/2))
```

```python
fig, axs = plt.subplots(2,3,figsize=(15,10)) 

fig.subplots_adjust(left=0.08,right=0.98,wspace=0.30,top=0.79,bottom=0.14)  

axs[0,0].plot(nu,W0_analyt,c='darkblue',lw=2,label='CMB') 
axs[0,0].plot(nu,W0,c='red',lw=2,label='CMB') 
axs[0,0].grid()
axs[0,0].set_ylabel(r'$W_{0}$',fontsize=20) 
axs[0,0].set_xlabel(r'$\nu$',fontsize=20) 

axs[0,1].plot(nu,W1_analyt,c='darkblue',lw=2,label='CMB') 
axs[0,1].plot(nu,W1,c='red',lw=2,label='CMB') 
axs[0,1].grid()
axs[0,1].set_ylabel(r'$W_{1}$',fontsize=20) 
axs[0,1].set_xlabel(r'$\nu$',fontsize=20) 

axs[0,2].plot(nu,W2_analyt,c='darkblue',lw=2,label='CMB') 
axs[0,2].plot(nu,W2,c='red',lw=2,label='CMB') 
axs[0,2].grid()
axs[0,2].set_ylabel(r'$W_{2}$',fontsize=20) 
axs[0,2].set_xlabel(r'$\nu$',fontsize=20) 

axs[1,0].plot(nu,W0-W0_analyt,c='orangered',lw=2,label='CMB') 
axs[1,0].grid()
axs[1,0].set_ylabel(r'$W_{1}^{0,2}$',fontsize=20) 
axs[1,0].set_xlabel(r'$\nu$',fontsize=20) 

axs[1,1].plot(nu,W1-W1_analyt,c='orangered',lw=2,label='CMB') 
axs[1,1].grid()
axs[1,1].set_ylabel(r'$W_{1}^{0,2}$',fontsize=20) 
axs[1,1].set_xlabel(r'$\nu$',fontsize=20) 

axs[1,2].plot(nu,W2-W2_analyt,c='orangered',lw=2,label='CMB') 
axs[1,2].grid()
axs[1,2].set_ylabel(r'$\alpha$',fontsize=20) 
axs[1,2].set_xlabel(r'$\nu$',fontsize=20) 

```

### At different angular scales.

We use the bandpass filter:
$$
f(\ell)=\frac{1}{4}\Big\{1+{\rm tanh}\Big(\frac{\ell-\ell_{c}}{\Delta \ell}\Big)\Big\}\Big\{1-{\rm tanh}\Big(\frac{\ell-\ell^{*}}{\Delta \ell}\Big)\Big\}
$$

```python
def apply_bp(mapA,ellc):
    alm=hp.map2alm(mapA,lmax) 
    dl=10.0   
    for ell in np.arange(lmax):
        xx=(ell-ellc)/dl
        yy=(ell-180)/dl   
        if ((ellc)==0):
            bp_ell=0.5*(1-tanh(yy))  
        else: 
            bp_ell=0.5*(1+tanh(xx))*0.5*(1-tanh(yy))
            
        for m in np.arange(ell+1):
            alm[int(hp.Alm.getidx(lmax,ell,m))]*=bp_ell 
        mapB=hp.alm2map(alm,nside,lmax) 
    return mapB     
```

```python
mask80=np.ones(npix)
mask80[haslam>80]=0

haslam20=apply_bp(haslam*mask80,20.0)
haslam50=apply_bp(haslam*mask80,50.0)
haslam90=apply_bp(haslam*mask80,90.0)
```

```python
hp.mollview(haslam,cmap='turbo',norm='hist')
hp.mollview(haslam20,cmap='turbo',norm='hist')
hp.mollview(haslam50,cmap='turbo',norm='hist')
hp.mollview(haslam90,cmap='turbo',norm='hist')
```

```python
haslam20=haslam20*has_mask
haslam20=(haslam20-np.mean(haslam20[has_mask>snu]))/(np.std(haslam20[has_mask>snu])) 

haslam50=haslam50*has_mask
haslam50=(haslam50-np.mean(haslam50[has_mask>snu]))/(np.std(haslam50[has_mask>snu])) 

haslam90=haslam90*has_mask
haslam90=(haslam90-np.mean(haslam90[has_mask>snu]))/(np.std(haslam90[has_mask>snu])) 


```
